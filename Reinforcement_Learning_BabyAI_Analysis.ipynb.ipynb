# **7.5 MDP Formulation**

# Markov Decision Process Formulation

A Reinforcement Learning problem can be represented using a Markov Decision Process (MDP).

An MDP is represented by:

**MDP = (S, A, P, R, γ)**

Where:

* **S** = set of states
* **A** = set of possible actions
* **P** = transition probability
* **R** = reward function
* **γ** = discount factor

For BabyAI-FindObjS5:

**State (S):** Current situation/observation of the agent.

**Action (A):** One of the available actions that the agent can perform.

**Transition (P):** The environment changes after the agent performs an action.

**Reward (R):** The agent receives feedback from the environment.

**Discount factor (γ):** Determines the importance of future rewards.

The objective is to learn a policy that maximizes the expected cumulative discounted reward.


# **7.7 Discounted Return**

## Return

The return represents the total discounted reward obtained by the agent.

The return is given by:

**Gₜ = Rₜ₊₁ + γRₜ₊₂ + γ²Rₜ₊₃ + ...**

where:

* **Gₜ** = return at time t
* **R** = reward
* **γ** = discount factor

A value of γ close to 1 gives more importance to future rewards, while a smaller γ gives more importance to immediate rewards.

For this project, the return will be used while implementing Monte Carlo, TD learning, SARSA and Q-Learning.

# **9.1 State Representation — Markdown Cell**

# State Representation

The BabyAI-FindObjS5 environment provides partial observations to the agent.

For tabular Reinforcement Learning algorithms such as Dynamic Programming, Monte Carlo, TD(0), SARSA and Q-Learning, a compact state representation is required.

In this project, the observation is converted into a discrete representation that can be used for tabular analysis.

The state representation is based on the information available in the observation rather than assuming access to the complete environment.

This representation allows us to demonstrate classical Reinforcement Learning algorithms while respecting the partial-observation nature of the problem.


# **9.2 Create a State Encoding Function**

```python
# Convert an observation into a compact hashable state representation

def encode_observation(observation):
    """
    Converts the observation into a tuple that can be
    used as a dictionary key in tabular RL algorithms.
    """

    encoded_parts = []

    # Process every observation component
    for key in sorted(observation.keys()):

        value = observation[key]

        # Numerical observations
        if isinstance(value, np.ndarray):
            encoded_parts.append(
                (key, value.tobytes())
            )

        # Text observations such as mission
        else:
            encoded_parts.append(
                (key, str(value))
            )

    return tuple(encoded_parts)
```

# **9.3 Convert Dataset Episodes into State-Action-Reward Data**

```python
# Convert the expert trajectories into state-action-reward transitions

transitions = []

for ep in dataset.iterate_episodes():

    observations = ep.observations
    actions = ep.actions
    rewards = ep.rewards

    # Number of transitions
    n = len(actions)

    for t in range(n):

        # Construct observation at time t
        obs_t = {
            key: observations[key][t]
            for key in observations.keys()
        }

        state = encode_observation(obs_t)

        action = int(actions[t])
        reward = float(rewards[t])

        transitions.append(
            (state, action, reward)
        )

print("Total transitions:", len(transitions))
print("First transition:")
print(transitions[0])
```

# **9.5 Value Function**

# State-Value and Action-Value Functions

## State-Value Function

The state-value function represents the expected return when the agent starts from a particular state and follows a policy.

It is represented as:

**Vπ(s) = Eπ[Gₜ | Sₜ = s]**

where:

* **Vπ(s)** = value of state s
* **Gₜ** = return from time t
* **π** = policy
* **E** = expected value

A state with a higher value is expected to provide a higher future return.

## Action-Value Function

The action-value function represents the expected return when the agent takes action `a` in state `s` and then follows a policy.

It is represented as:

**Qπ(s,a) = Eπ[Gₜ | Sₜ = s, Aₜ = a]**

Q-values are particularly important for algorithms such as SARSA and Q-Learning.

# **10.1 Bellman Equation — Markdown Cell**

# Bellman Equations

Bellman equations are fundamental equations in Reinforcement Learning. They express the value of a state in terms of the immediate reward and the value of the next state.

## Bellman Expectation Equation

For a policy π:

**Vπ(s) = Σₐ π(a|s) Σₛ′ P(s′|s,a)[R(s,a,s′) + γVπ(s′)]**

For the action-value function:

**Qπ(s,a) = Σₛ′ P(s′|s,a)[R(s,a,s′) + γ Σₐ′π(a′|s′)Qπ(s′,a′)]**

## Bellman Optimality Equation

For the optimal state-value function:

**V*(s) = maxₐ Σₛ′ P(s′|s,a)[R + γV*(s′)]**

For the optimal action-value function:

**Q*(s,a) = Σₛ′ P(s′|s,a)[R + γ maxₐ′Q*(s′,a′)]**

These equations are used by Dynamic Programming and Q-Learning to update value estimates.


# **10.2 Create an Empirical Transition Model**

```python
# Build state-action-next-state-reward transitions

sa_transitions = []

for ep in dataset.iterate_episodes():

    observations = ep.observations
    actions = ep.actions
    rewards = ep.rewards

    states = []

    # Encode every observation in this episode
    for t in range(len(actions)):
        obs_t = {
            key: observations[key][t]
            for key in observations.keys()
        }

        states.append(encode_observation(obs_t))

    # Create transitions
    for t in range(len(actions)):

        state = states[t]
        action = int(actions[t])
        reward = float(rewards[t])

        # Last transition has no next state
        if t < len(actions) - 1:
            next_state = states[t + 1]
        else:
            next_state = None

        sa_transitions.append(
            (state, action, reward, next_state)
        )

print("Total state-action transitions:", len(sa_transitions))
```

# **10.3 Build the Transition Model**

```python
from collections import defaultdict, Counter

# Count observed next states for every state-action pair

transition_counts = defaultdict(Counter)
reward_values = defaultdict(list)

for state, action, reward, next_state in sa_transitions:

    if next_state is not None:
        transition_counts[(state, action)][next_state] += 1

    reward_values[(state, action)].append(reward)


print("Number of state-action pairs:",
      len(transition_counts))
```

```python
# Convert transition counts into empirical probabilities

transition_model = {}

for key, counts in transition_counts.items():

    total = sum(counts.values())

    transition_model[key] = {
        next_state: count / total
        for next_state, count in counts.items()
    }

print("Empirical transition model created.")
```

# **10.4 Dynamic Programming Explanation**

# Dynamic Programming

Dynamic Programming methods use the Bellman equations to estimate values of states or actions.

Dynamic Programming normally requires a model of the environment, including transition probabilities and rewards.

Since our project uses an offline expert-trajectory dataset, the exact environment model is not learned directly from the dataset. Instead, an empirical transition model is estimated from the observed transitions.

We can then use this empirical model for a small value-iteration demonstration.

The goal is to repeatedly update the value of each state until the values become stable.


# **10.5 Value Iteration**

```python
# Value Iteration using the empirical transition model

gamma = 0.95
theta = 1e-6
max_iterations = 100

# Initialize all state values to zero
V = {state: 0.0 for state in unique_states}

states = list(unique_states)

for iteration in range(max_iterations):

    delta = 0.0

    new_V = V.copy()

    for state in states:

        action_values = []

        # Find actions observed from this state
        available_actions = {
            action
            for s, action, reward, next_state in sa_transitions
            if s == state
        }

        for action in available_actions:

            key = (state, action)

            if key not in transition_model:
                continue

            expected_value = 0.0

            # Calculate Bellman update
            for next_state, probability in transition_model[key].items():

                rewards = reward_values[key]
                reward = np.mean(rewards)

                expected_value += probability * (
                    reward + gamma * V.get(next_state, 0.0)
                )

            action_values.append(expected_value)

        if action_values:

            best_value = max(action_values)

            delta = max(
                delta,
                abs(best_value - V[state])
            )

            new_V[state] = best_value

    V = new_V

    if delta < theta:
        print("Value iteration converged at iteration:", iteration + 1)
        break

else:
    print("Maximum iterations reached.")

print("Number of states with estimated values:", len(V))
```

# **11.1 Theory — Markdown Cell**

# Monte Carlo Method

Monte Carlo methods estimate the value of states by using the actual returns obtained from complete episodes.

Unlike Dynamic Programming, Monte Carlo methods do not require a complete transition model of the environment.

For a state visited at time t, the return is:

**Gₜ = Rₜ₊₁ + γRₜ₊₂ + γ²Rₜ₊₃ + ...**

The state value is estimated by averaging the returns observed after visiting that state.

## First-Visit Monte Carlo

In First-Visit Monte Carlo, the return is calculated only for the first occurrence of a state in an episode.

The estimated value is:

**V(s) = Average of returns observed after visiting s**

Monte Carlo learning is useful for our offline dataset because complete expert episodes are available.


# **11.2 Prepare Episodes**

```python
# Store all episodes in a convenient format

episodes_data = []

for ep in dataset.iterate_episodes():

    observations = ep.observations
    actions = ep.actions
    rewards = ep.rewards

    states = []

    for t in range(len(actions)):

        obs_t = {
            key: observations[key][t]
            for key in observations.keys()
        }

        states.append(encode_observation(obs_t))

    episodes_data.append({
        "states": states,
        "actions": actions,
        "rewards": rewards
    })

print("Episodes prepared:", len(episodes_data))
```

# **11.3 Implement First-Visit Monte Carlo**

```python
from collections import defaultdict

gamma = 0.95

# Store returns for every state
mc_returns = defaultdict(list)

for episode_data in episodes_data:

    states = episode_data["states"]
    rewards = episode_data["rewards"]

    visited_states = set()

    # Calculate return backwards
    G = 0.0

    for t in reversed(range(len(states))):

        G = float(rewards[t]) + gamma * G

        state = states[t]

        # First-visit condition
        if state not in visited_states:

            mc_returns[state].append(G)

            visited_states.add(state)


# Average return for each state
MC_V = {
    state: np.mean(returns)
    for state, returns in mc_returns.items()
}

print("Monte Carlo states estimated:", len(MC_V))
```

# **12.1 Theory — Markdown Cell**

# Temporal Difference Learning

Temporal Difference (TD) learning is a Reinforcement Learning method that learns from experience without waiting for the complete episode to finish.

TD learning combines ideas from Dynamic Programming and Monte Carlo methods.

It updates the value of a state using:

* Current state
* Immediate reward
* Next state
* Current estimate of the next state's value

## TD(0) Update

The TD(0) update rule is:

**V(Sₜ) ← V(Sₜ) + α[Rₜ₊₁ + γV(Sₜ₊₁) − V(Sₜ)]**

Where:

* **V(Sₜ)** = current value of the state
* **α** = learning rate
* **Rₜ₊₁** = immediate reward
* **γ** = discount factor
* **V(Sₜ₊₁)** = estimated value of the next state

The term:

**Rₜ₊₁ + γV(Sₜ₊₁)**

is called the TD target.

The difference between the TD target and current value is called the **TD error**.


# **12.2 TD(0) Implementation**

```python
from collections import defaultdict

# TD(0) parameters
alpha = 0.1
gamma = 0.95

# Initialize state values
TD_V = defaultdict(float)

# Store average TD error for analysis
td_errors = []

# Run TD learning over the available episodes
for episode_data in episodes_data:

    states = episode_data["states"]
    rewards = episode_data["rewards"]

    for t in range(len(states) - 1):

        state = states[t]
        next_state = states[t + 1]
        reward = float(rewards[t])

        # TD target
        td_target = reward + gamma * TD_V[next_state]

        # TD error
        td_error = td_target - TD_V[state]

        # TD(0) update
        TD_V[state] += alpha * td_error

        td_errors.append(td_error)

print("TD(0) learning completed.")
print("Number of states:", len(TD_V))
```

TD(0) updates the value of a state using the immediate reward and the estimated value of the next state.

# **13.1 SARSA Theory — Markdown Cell**

# SARSA

SARSA is an **on-policy Temporal Difference learning algorithm**.

The name SARSA comes from the sequence:

**State → Action → Reward → Next State → Next Action**

SARSA updates the action-value function using:

**Q(Sₜ,Aₜ) ← Q(Sₜ,Aₜ) + α[Rₜ₊₁ + γQ(Sₜ₊₁,Aₜ₊₁) − Q(Sₜ,Aₜ)]**

where:

* **Q(S,A)** = action-value function
* **α** = learning rate
* **γ** = discount factor
* **R** = reward
* **S** = state
* **A** = action

SARSA is called on-policy because it updates the value of the action that the current policy actually selects.


# **13.2 Prepare SARSA Data**

```python
# Prepare state-action-next-action transitions for SARSA

sarsa_transitions = []

for episode_data in episodes_data:

    states = episode_data["states"]
    actions = episode_data["actions"]
    rewards = episode_data["rewards"]

    for t in range(len(actions) - 1):

        state = states[t]
        action = int(actions[t])
        reward = float(rewards[t])

        next_state = states[t + 1]
        next_action = int(actions[t + 1])

        sarsa_transitions.append(
            (state, action, reward, next_state, next_action)
        )

print("SARSA transitions:", len(sarsa_transitions))
```

# **13.3 Implement SARSA**

```python
from collections import defaultdict

# SARSA parameters
alpha = 0.1
gamma = 0.95

# Initialize Q-values
Q_sarsa = defaultdict(float)

# Store TD errors
sarsa_errors = []

# Train using the observed expert trajectories
for state, action, reward, next_state, next_action in sarsa_transitions:

    current_q = Q_sarsa[(state, action)]

    next_q = Q_sarsa[(next_state, next_action)]

    # SARSA target
    target = reward + gamma * next_q

    # TD error
    error = target - current_q

    # SARSA update
    Q_sarsa[(state, action)] += alpha * error

    sarsa_errors.append(error)

print("SARSA learning completed.")
print("State-action pairs learned:", len(Q_sarsa))
```

# **14.1 Q-Learning Theory — Markdown Cell**

# Q-Learning

Q-Learning is an **off-policy Temporal Difference learning algorithm**.

It learns the optimal action-value function **Q*(s,a)**.

The Q-Learning update rule is:

**Q(Sₜ,Aₜ) ← Q(Sₜ,Aₜ) + α[Rₜ₊₁ + γ maxₐ Q(Sₜ₊₁,a) − Q(Sₜ,Aₜ)]**

where:

* **Q(S,A)** = action-value function
* **α** = learning rate
* **γ** = discount factor
* **R** = reward
* **max Q(S',a)** = maximum estimated value of the available next actions

Q-Learning is called off-policy because it learns toward the best estimated next action rather than necessarily using the action observed in the trajectory.


# **14.2 Prepare Q-Learning Transitions**

```python
# Prepare transitions for Q-Learning

qlearning_transitions = []

for episode_data in episodes_data:

    states = episode_data["states"]
    actions = episode_data["actions"]
    rewards = episode_data["rewards"]

    for t in range(len(actions) - 1):

        state = states[t]
        action = int(actions[t])
        reward = float(rewards[t])
        next_state = states[t + 1]

        qlearning_transitions.append(
            (state, action, reward, next_state)
        )

print("Q-Learning transitions:", len(qlearning_transitions))
```

# **14.3 Implement Q-Learning**

```python
from collections import defaultdict

# Q-Learning parameters
alpha = 0.1
gamma = 0.95

# Initialize Q-values
Q_learning = defaultdict(float)

# Store TD errors
qlearning_errors = []

# Train on the observed transitions
for state, action, reward, next_state in qlearning_transitions:

    current_q = Q_learning[(state, action)]

    # Find actions observed for the next state
    next_actions = [
        a
        for s, a, r, ns in qlearning_transitions
        if s == next_state
    ]

    if next_actions:

        max_next_q = max(
            Q_learning[(next_state, a)]
            for a in set(next_actions)
        )

    else:
        max_next_q = 0.0

    # Q-Learning target
    target = reward + gamma * max_next_q

    # TD error
    error = target - current_q

    # Update Q-value
    Q_learning[(state, action)] += alpha * error

    qlearning_errors.append(error)

print("Q-Learning completed.")
print("State-action pairs learned:", len(Q_learning))
```

# **14.7 SARSA vs Q-Learning**

# SARSA vs Q-Learning

SARSA and Q-Learning are both Temporal Difference control algorithms, but they use different update rules.

| Feature     | SARSA                              | Q-Learning                             |
| ----------- | ---------------------------------- | -------------------------------------- |
| Type        | On-policy                          | Off-policy                             |
| Learns      | Q(s,a)                             | Q(s,a)                                 |
| Next action | Uses observed/selected next action | Uses maximum estimated next Q-value    |
| Update      | Uses Q(s′,a′)                      | Uses max Q(s′,a′)                      |
| Main idea   | Learn the behavior being followed  | Learn toward the best estimated action |

### SARSA Update

**Q(s,a) ← Q(s,a) + α[r + γQ(s′,a′) − Q(s,a)]**

### Q-Learning Update

**Q(s,a) ← Q(s,a) + α[r + γ max Q(s′,a′) − Q(s,a)]**

# **15.1 Function Approximation — Markdown Cell**

# Function Approximation in Reinforcement Learning

In tabular Reinforcement Learning, a separate value is stored for every state or state-action pair.

However, environments with image observations can contain a very large number of possible states. Storing a separate table entry for every state becomes inefficient.

Function approximation solves this problem by using a parameterized function, such as a neural network, to estimate value functions.

Instead of storing:

**Q(s,a)**

for every state-action pair, a neural network can learn:

**Q(s,a; θ)**

where **θ** represents the neural network parameters.

The network learns general patterns from observations and can estimate Q-values for states that were not explicitly stored in a table.

The BabyAI-FindObjS5 environment provides partial observations that include visual information.

A neural network can process the observation and approximate the action-value function.

# **16.1 Policy Gradient — Markdown Cell**

# Policy Gradient Methods

Policy Gradient methods directly learn a policy instead of first learning a Q-table.

A policy represents the probability of selecting an action given a state:

**π(a|s)**

A neural network can be used to represent the policy.

The network receives the state as input and produces probabilities for the available actions.

The objective is to maximize the expected return:

**J(θ) = Eπθ[Gₜ]**

where **θ** represents the parameters of the policy network.

Policy Gradient methods are useful for environments with large or continuous state or action spaces.


# **16.2 Policy Network**

```python
from tensorflow.keras.layers import Softmax

# Policy network
policy_network = Sequential([
    Dense(64, activation="relu", input_shape=(10,)),
    Dense(64, activation="relu"),
    Dense(num_actions, activation="softmax")
])

policy_network.summary()
```

# **16.4 REINFORCE — Markdown Cell**

# REINFORCE Algorithm

REINFORCE is a Monte Carlo Policy Gradient algorithm.

It learns a policy directly by using the returns obtained from complete episodes.

The policy parameters are updated using:

**θ ← θ + α Gₜ ∇θ log πθ(Aₜ|Sₜ)**

where:

* **θ** = policy parameters
* **α** = learning rate
* **Gₜ** = return from time t
* **πθ(Aₜ|Sₜ)** = probability of the selected action
* **∇** = gradient

The basic steps of REINFORCE are:

1. Generate an episode.
2. Calculate returns.
3. Calculate action probabilities.
4. Increase the probability of actions that produced good returns.
5. Update the policy parameters.


# **16.5 Simple REINFORCE Demonstration**

```python
# Demonstrate the REINFORCE loss calculation

sample_action = 0
sample_return = 1.0

probability = action_probabilities[sample_action]

# Avoid log(0)
probability = max(probability, 1e-8)

reinforce_loss = -sample_return * np.log(probability)

print("Selected action:", sample_action)
print("Action probability:", probability)
print("Return:", sample_return)
print("REINFORCE loss:", reinforce_loss)
```

# **16.6 Actor–Critic — Markdown Cell**

# Actor–Critic Method

Actor–Critic combines Policy Gradient and Value Function methods.

It contains two main components:

### Actor

The Actor learns the policy.

It decides which action should be selected.

### Critic

The Critic estimates the value of the current state.

It evaluates how good the current situation is.

The TD error can be used to guide the Actor:

**δ = r + γV(s′) − V(s)**

The Actor updates the policy using this learning signal, while the Critic updates the value function.

Therefore:

**Actor → selects actions**

**Critic → evaluates actions/states**
